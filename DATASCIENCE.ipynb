```python
arr1=np.array([1,2,3])

arr2=np.array([3,2,4])

print(arr1+arr2)

print(arr1-arr2)

print(arr1*arr2)

print(arr1/arr2)

print(arr1**arr2)

print(arr1%arr2)
```

```python
arr1=np.array([1,2,3])

arr2=np.array([3,2,4])

print(arr1>arr2)

print(arr1==arr2)

print(arr1!=arr2)
```
