```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)   #draw graph
plt.show()      #dispay graph
```

```python
months=["Nov","Dec","Jan","Feb","Mar"]
sales=[79,30,75,80,40]
plt.plot(months,sales)
plt.show()
```

```python
months=["Nov","Dec","Jan","Feb","Mar"]
sales=[79,30,75,80,40]
profit=[6,8,11,14,15]
plt.plot(months,sales,label="sales")
plt.plot(months,profit,label="profit")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["Nov","Dec","Jan","Feb","Mar"]
sales=[40,30,75,80,35]
plt.bar(months,sales)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.grid("True")
plt.plot(
    months,
    sales,
    marker="*",
    color="red",
    linestyle="--"
)
plt.show()
```

```python
months=["Nov","Dec","Jan","Feb","Mar"]
sales=[79,30,75,80,40]
profit=[6,8,11,14,15]
plt.plot(months,sales,label="sales")
plt.plot(months,profit,label="profit")
plt.legend()
plt.show()
```

```python
marks=[33,55,80,34,50,66,83,50,78,95]
plt.hist(marks)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
categories=["Food","Travel","Shopping","Others"]
amount=[6000,7000,10000,3000]
plt.pie(amount,labels=categories)
plt.title("Monthly Expenses")
plt.show()
```
