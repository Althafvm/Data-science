```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data={
    "product":["Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
    "Sales":[50,80,40,70,90,60],
    "Price":[50000,30000,20000,55000,32000,22000],
    "Month":["Jan","Jan","Feb","Feb","Mar","mar"]
}
sns.barplot(x="Month",y="Sales",data=data)
plt.show()
```

```python
sns.histplot(x="Sales",data=data)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data=np.random.randn(1000)
plt.figure(figsize=(8,6))
sns.histplot(data,bins=30,kde=True,color="skyblue",edgecolor="black")
plt.xlabel("values")
plt.ylabel("frequency")
plt.title("distribution of random data using seaborn")
plt.grid(axis="y",alpha=0.75)
plt.show()
```

```python
months=["jan","fab","mar","apr"]
sales=[20,30,40,50]
plt.plot(months,sales,marker="^",linestyle="-.")
plt.xlabel("moths")
plt.ylabel("sales")
plt.title("sales per month")
plt.grid(True)
plt.show()
```

```python
products=["pen","ben","men","zen"]
sales=[20,30,40,50]
plt.barh(products,sales)
plt.xlabel("products")
plt.ylabel("sales")
plt.title("product sales")
plt.show()
```

```python
products=["laptop","phone","tablet"]
sales=[50,60,70]
plt.plot(sales,labels=products,autopct="%")
```

```python
x=[3,5,7,8,9]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.fill_between(x,y)
plt.show()
```

```python
marks=[34,56,78,85,90,55,67,89]
sns.boxplot(x=marks)
plt.show()
```

```python
sales=[
    [10,20,30,40],
    [15,25,35,45],
    [20,30,40,50]
]
sns.heatmap(sales,annot=True,cmap="Purples")
plt.show()
```
