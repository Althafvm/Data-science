```python
import pandas as pd
import numpy as np
data={"name":["alice","bob","charlie","david","emma"],"age":[20,21,19,22,20],"marks":[85,72,np.nan,90,65],"attendance":[95,80,75,np.nan,88],"city":["kochi","delhi","munbai","kochi","chennai"]}
df=pd.DataFrame(data)
print("original dataframe")
print(df)
print("\nFirst 3 rows:")
print(df.head(2))

print("\nFirst 2 rows:")
print(df.tail(3))

print("\nDataFrame shape")
print(df.shape)

print("\ncolumns name")
print(df.columns)

print("\ndatatype name")
print(df.dtypes)

print("\nmarks columns")
print(df.marks)

print("\nname and marks")
print(df[["name","marks"]])

print("\nfrist 3 rows")
print(df.iloc[0:3])

print("\nrow 1 to 3, name and marks:")
print(df.loc[1:3,["name","marks"]])

print("\nrow 1 to 3, name,marks,age,city and attendance:")
print(df.loc[1:3,["name","marks","age","city","attendance"]])

print("\nstudent with marks greater than 75:")
print(df[df["marks"]>75])

print("\nstudent from kochi:")
print(df[df["city"]== "kochi"])

print("\nstudent with marks>70 AND attendance>80:")
print(df[
    (df["marks"]>70) &
    (df["attendance"]>80)
    ])
print("\nmarks +5:")
print(df["marks"]+5)

print("\nmarks *2:")
print(df["marks"]*2)

print("\naverage marks:")
print(df["marks"].mean())

print("\nmaximum marks:")
print(df["marks"].max())

print("\nminimum marks:")
print(df["marks"].min())

print("\ntotal marks:")
print(df["marks"].sum())

print("\nsorted by marks:")
print(df.sort_values("marks"))

print("\nsorted by marks - highest frist:")
print(df.sort_values("marks",ascending=False))
```

```python
df_renamed =df.rename(
    columns={
        "marks":"score",
        "attendance":"attendance_percentage"
    }
)
print("\nrenamed dataframe:")
print(df_renamed)
```

```python
print("\nmissing values:")
print(df.isna())

print("\nnumber of missing values:")
print(df.isna().sum())
```

```python
#forward fill
df_forward =df.copy()
df_forward = df_forward.ffill()
print("\nforward fill:")
print(df_forward)
```

```python
#backward fill
df_backward =df.copy()
df_backward = df_backward.bfill()
print("\nbackward fill:")
print(df_backward)
```
